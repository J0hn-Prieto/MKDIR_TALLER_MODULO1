### 1.1 — Listas y operaciones básicas

Tienes la siguiente lista con la esperanza de vida de 10 países (en años).

````python
# Lista dada — NO modificar
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]
````

**Ejercicio 1.1** — Usando Python puro (sin Pandas ni funciones built-in como `sum()` o `max()`):
- Calcula el **promedio** de la lista usando un bucle `for`
- Encuentra el **valor máximo** usando un bucle `for`
- Encuentra el **valor mínimo** usando un bucle `for`

````python
# Ejercicio 1.1
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]

# Promedio
total = 0
for valor in esperanzas_de_vida:
    total = total + valor

cantidad = 0
for _ in esperanzas_de_vida:
    cantidad = cantidad + 1

promedio = total / cantidad

print(f"Suma total:         {total}")
print(f"Cantidad de países: {cantidad}")
print(f"Promedio:           {promedio}")

# Máximo
maximo = esperanzas_de_vida[0]
for valor in esperanzas_de_vida:
    if valor > maximo:
        maximo = valor

print(f"Valor máximo: {maximo}")

# Mínimo
minimo = esperanzas_de_vida[0]
for valor in esperanzas_de_vida:
    if valor < minimo:
        minimo = valor

print(f"Valor mínimo: {minimo}")
````

### 1.3 — Funciones

Las funciones nos permiten reutilizar lógica.

**Ejercicio 1.3** — Crea una función llamada `clasificar_esperanza_vida` que reciba un número (años) y retorne:
- `"Alta"` si es >= 75
- `"Media"` si es entre 60 y 74.9
- `"Baja"` si es < 60

Prueba la función con al menos 5 valores distintos e imprime los resultados.

````python
# Ejercicio 1.3

def clasificar_esperanza_vida(años):
    if años >= 75:
        return "Alta"
    elif años >= 60:
        return "Media"
    else:
        return "Baja"

# Pruebas
print(clasificar_esperanza_vida(83.2))
print(clasificar_esperanza_vida(67.8))
print(clasificar_esperanza_vida(49.3))
print()
# Prueba con 5 valores diferentes
print(clasificar_esperanza_vida(96.6))
print(clasificar_esperanza_vida(65.5))
print(clasificar_esperanza_vida(88.8))
print(clasificar_esperanza_vida(22.0))
print(clasificar_esperanza_vida(19.1))

````

### 2.3 — Estadísticas descriptivas

🔧 **Ejercicio 2.3**
1. Muestra el resumen estadístico completo del dataset
2. ¿Cuál es la esperanza de vida **promedio** mundial?
3. ¿Cuál es el país con **mayor** gasto en salud? ¿Y el de **menor**?
4. ¿Cuál es la **mediana** de la mortalidad infantil?

````python
# Ejercicio 2.3

# 1. Muestra el resumen estadístico completo del dataset
print(df.describe())

# 2. ¿Cuál es la esperanza de vida **promedio** mundial?
promedio_ev = df["esperanza_vida"].mean()
print("-"*50)
print(f"Esperanza de vida promedio mundial: {promedio_ev:.2f} años")

# 3. ¿Cuál es el país con **mayor** gasto en salud? ¿Y el de **menor**?
mayor_gasto = df.loc[df["gasto_salud_usd"].idxmax(), "pais"]
menor_gasto = df.loc[df["gasto_salud_usd"].idxmin(), "pais"]

print(f"Pais con mayor gasto en salud: {mayor_gasto}")
print(f"Pais con menor gasto en salud: {menor_gasto}")

# 4. ¿Cuál es la **mediana** de la mortalidad infantil?
mediana_mortalidad = df["mortalidad_infantil"].median()
print(f"Mediana de la mortalidad infantil: {mediana_mortalidad:.2f}")

````

### 2.4 — Filtrado y selección

🔧 **Ejercicio 2.4**
1. Filtra los países de **América** (`region == "Americas"`) y muestra solo las columnas: `pais`, `esperanza_vida`, `gasto_salud_usd`
2. Filtra los países con esperanza de vida **mayor a 80 años** — ¿cuántos son? ¿De qué regiones son?
3. Filtra los países con **nivel_ingresos == "Low"** y calcula su promedio de `mortalidad_infantil`
4. ¿Cuántos países tienen vacunación por encima del 90%?

````python
# 2.4 Filtrado y seleccion

# 1. Filtra los países de **América** (`region == "Americas"`) y muestra solo las columnas: `pais`, `esperanza_vida`, `gasto_salud_usd`
americas = df[df["region"] == "Americas"][["pais", "esperanza_vida", "gasto_salud_usd"]]
print(americas)
print("-"*60)

# 2. Filtra los países con esperanza de vida **mayor a 80 años** — ¿cuántos son? ¿De qué regiones son?
ev_mayor_80 = df[df["esperanza_vida"] > 80]
print(f"Cantidad de países con esperanza de vida: {len(ev_mayor_80)}")
print(ev_mayor_80[["pais", "region", "esperanza_vida"]])
print("-"*55)

# 3. Filtra los países con **nivel_ingresos == "Low"** y calcula su promedio de `mortalidad_infantil`
low_income = df[df["nivel_ingresos"] == "Low"]
promedio_mortalidad = low_income["mortalidad_infantil"].mean()
print(f"Promedio de mortalidad infantil (Low): {promedio_mortalidad:.2f}")
print("-"*55)

# 4. ¿Cuántos países tienen vacunación por encima del 90%?
vacunacion_90 = df[df["vacunacion_pct"] > 90]
print(f"Países con vacunación por encima del 90%: {len(vacunacion_90)}")
print("-"*55)
````

### 2.5 — Agrupaciones

🔧 **Ejercicio 2.5** — Usa `groupby()` para:
1. Calcular la **esperanza de vida promedio** por `region`
2. Calcular el **gasto en salud promedio** por `nivel_ingresos`
3. Calcular el **número de países** por región
4. ¿Qué región tiene la mayor esperanza de vida promedio? ¿Y la menor?

````python
# 2.5 Agrupaciones

# 1. Calcular la **esperanza de vida promedio** por `region`
ev_por_region = df.groupby("region")["esperanza_vida"].mean()
print(ev_por_region)
print("-"*40)

# 2. Calcular el **gasto en salud promedio** por `nivel_ingresos`
gasto_por_ingresos = df.groupby("nivel_ingresos")["gasto_salud_usd"].mean()
print(gasto_por_ingresos)
print("-"*40)

# 3. Calcular el **número de países** por región
paises_por_region = df.groupby("region")["pais"].count()
print(paises_por_region)
print("-"*40)

# 4. ¿Qué región tiene la mayor esperanza de vida promedio? ¿Y la menor?
region_mayor = ev_por_region.idxmax()
region_menor = ev_por_region.idxmin()

print(f"Region con mayor esperanza de vida: {region_mayor} ({ev_por_region[region_mayor]:.2f} años)")
print(f"Region con menor esperanza de vida: {region_menor} ({ev_por_region[region_menor]:.2f} años)")

````

### 2.6 — Nueva columna calculada

🔧 **Ejercicio 2.6**
1. Crea una nueva columna llamada `categoria_vida` aplicando la función `clasificar_esperanza_vida` que creaste en la Parte 1 a la columna `esperanza_vida`
2. ¿Cuántos países caen en cada categoría? (usa `value_counts()`)
3. ¿Qué porcentaje del total representa cada categoría?

````python
# 2.6 Nueva Columna calculada

# 1. Crea una nueva columna llamada `categoria_vida` aplicando la función `clasificar_esperanza_vida` que creaste en la Parte 1 a la columna `esperanza_vida`
df["categoria_vida"] = df["esperanza_vida"].apply(clasificar_esperanza_vida)
print(df[["pais", "esperanza_vida", "categoria_vida"]].head(10))
print("-"*45)

# 2. ¿Cuántos países caen en cada categoría? (usa `value_counts()`)
conteo = df["categoria_vida"].value_counts()
print(conteo)
print("-"*45)

# 3. ¿Qué porcentaje del total representa cada categoría?
porcentaje = df["categoria_vida"].value_counts(normalize=True) * 100
print(porcentaje.round(2))
````

**¿Cuál es la diferencia entre `git add .` y `git add nombre_archivo`?**

> Respuesta:

La diferencia está en el control Con git add . uno le dice a Git "toma todo lo que cambié", es rápido pero a veces descuidado si hay archivos que todavía no están listos.
